```python
#find the given number is prime or not
n = int(input())
for i in range(2,n):
  if(n%i == 0):
    print("it is not a prime")
    break
else:
  print("It is a prime number")
print("end")
```

```python
#print prime numbers between given two numbers
m = int(input())
n = int(input())
for i in range (m,n):
  for j in range(2,i):
    if(i%j == 0):
      break
  else:
      print(i,end=" ")

```
